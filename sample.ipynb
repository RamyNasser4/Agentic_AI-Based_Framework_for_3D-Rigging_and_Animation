```python
import numpy as np
import pandas as pd

file = r"CMU/CMU/01/01_01_poses.npz"
data = np.load(file, allow_pickle=True)

i = 0  # frame index
frame_pose = data["poses"][i]
frame_trans = data["trans"][i]
betas = data["betas"]

# Create a table for joint rotations
joint_indices = [f"Joint {j+1}" for j in range(len(frame_pose)//3)]
rotations = [frame_pose[j*3:(j+1)*3] for j in range(len(frame_pose)//3)]

df_poses = pd.DataFrame(rotations, index=joint_indices, columns=["Rx", "Ry", "Rz"])

# Root translation row
df_trans = pd.DataFrame([frame_trans], index=["Root Translation"], columns=["X", "Y", "Z"])

# Betas row
beta_indices = [f"Beta {j+1}" for j in range(len(betas))]
df_betas = pd.DataFrame([betas], columns=beta_indices)

# Display nicely
print("Joint Rotations (Axis–Angle) for Frame", i)
display(df_poses.round(3))
print("\nRoot Joint Translation")
display(df_trans.round(3))
print("\nBody Shape Parameters (Betas)")
display(df_betas.round(3))
```

```python
import numpy as np
import pandas as pd

# Load a sample frame
file = r"CMU/CMU/01/01_01_poses.npz"
data = np.load(file, allow_pickle=True)

i = 0  # frame index
frame_pose = data["poses"][i]
frame_trans = data["trans"][i]
betas = data["betas"]

# ===================== 1️⃣ Joint Rotations =====================
num_joints = len(frame_pose) // 3
joint_names = [f"Joint {j+1}" for j in range(num_joints)]
rotation_columns = ["Rx", "Ry", "Rz"]
rotations = [frame_pose[j*3:(j+1)*3] for j in range(num_joints)]

df_poses = pd.DataFrame(rotations, index=joint_names, columns=rotation_columns)

# Add multi-level header
df_poses.columns = pd.MultiIndex.from_product([["Joint Rotations"], rotation_columns])

# ===================== 2️⃣ Root Translation =====================
df_trans = pd.DataFrame([frame_trans], index=["Root Translation"], columns=["X", "Y", "Z"])
df_trans.columns = pd.MultiIndex.from_product([["Root Translation (m)"], df_trans.columns])

# ===================== 3️⃣ Body Shape (Betas) =====================
beta_names = [f"Beta {j+1}" for j in range(len(betas))]
df_betas = pd.DataFrame([betas], index=["Betas"])
df_betas.columns = pd.MultiIndex.from_product([["Body Shape Parameters"], df_betas.columns])

# ===================== Display =====================
print(f"Frame {i} - Joint Rotations, Translation, Betas")
display(df_poses.round(3))
display(df_trans.round(3))
display(df_betas.round(3))
```
